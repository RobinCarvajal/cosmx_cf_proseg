```python
MAIN_DIR = '/mnt/data/project0062/proseg_data'
os.chdir(MAIN_DIR)
os.getcwd()
```

```python
slide01_path = "data/slides/zarr/SLIDE01_proseg.zarr"
slide02_path = "data/slides/zarr/SLIDE02_proseg.zarr"
slide03_path = "data/slides/zarr/SLIDE03_proseg.zarr"
slide04_path = "data/slides/zarr/SLIDE04_proseg.zarr"
```

```python
sdata01 = sd.read_zarr(slide01_path)
sdata02 = sd.read_zarr(slide02_path)
sdata03 = sd.read_zarr(slide03_path)
sdata04 = sd.read_zarr(slide04_path)
```

```python
adata01 = sdata01['table'].copy()
adata02 = sdata02['table'].copy()
adata03 = sdata03['table'].copy()
adata04 = sdata04['table'].copy()
```

```python
adata01.write_h5ad("data/slides/h5ad/SLIDE01_proseg.h5ad")
adata02.write_h5ad("data/slides/h5ad/SLIDE02_proseg.h5ad")
adata03.write_h5ad("data/slides/h5ad/SLIDE03_proseg.h5ad")
adata04.write_h5ad("data/slides/h5ad/SLIDE04_proseg.h5ad")
```

```python
# save metadata
adata01_meta = adata01.obs.copy()
adata02_meta = adata02.obs.copy()
adata03_meta = adata03.obs.copy()
adata04_meta = adata04.obs.copy()
```

```python
adata01_meta.to_csv('data/slides/tables/SLIDE01_meta.tsv', sep='\t')
adata02_meta.to_csv('data/slides/tables/SLIDE02_meta.tsv', sep='\t')
adata03_meta.to_csv('data/slides/tables/SLIDE03_meta.tsv', sep='\t')
adata04_meta.to_csv('data/slides/tables/SLIDE04_meta.tsv', sep='\t')
```

The purpose of this noteboook is to substract the FOV numbers that make up a sample (after graphical selectio with SubsetMask)

```python
import pandas as pd
from pathlib import Path
import os
import json
```

```python
# %% setting paths
MAIN_DIR_NAME = "cosmx_gray"
MAIN_DIR = next(p for p in Path.cwd().parents if (p / MAIN_DIR_NAME).exists()) / MAIN_DIR_NAME
os.chdir(MAIN_DIR)
```

## Slide 1

```python
SLIDES_DIR = Path('data/slides')

slide_names = [
    d.name for d in SLIDES_DIR.iterdir() 
    if d.is_dir() and (
        "slide" in d.name.lower()
    )
]
print(f"Slides found: {slide_names}")
```

```python
for slide in slide_names:
    SLIDE_DIR = MAIN_DIR / f'data/slides/{slide}'
    meta_path = SLIDE_DIR / f'tables/{slide}_meta.tsv'
    
    # %% read metadata
    meta_df = pd.read_csv(meta_path, sep='\t')
    
    # %% get sample FOVs
    sample_names = meta_df['sample_name'].unique().tolist()

    sample_fovs = {}
    for sample in sample_names:
        sample_meta = meta_df[meta_df['sample_name'] == sample]
        fovs = sample_meta['fov'].unique().tolist()
        sample_fovs[sample] = fovs
    
    # save as json
    output_path = SLIDE_DIR / f'dicts/sample_fovs.json'
    output_path.parent.mkdir(parents=True, exist_ok=True)
    with open(output_path, 'w') as f:
        json.dump(sample_fovs, f)
    
    print(sample_fovs)
```
